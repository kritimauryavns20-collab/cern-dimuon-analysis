## Introduction

In high-energy particle physics, dimuon events occur when two muons are produced from particle collisions. These events are particularly important because they can provide evidence of fundamental particles such as the **Z boson**, which can decay into a pair of muons.

This project analyzes dimuon data from CERN to explore the properties of these events. By studying variables such as invariant mass and energy distributions, we aim to identify patterns that reveal underlying physical processes.

## Dataset Description

The dataset contains measurements from proton-proton collision events recorded at CERN. Each row represents a dimuon event, with variables describing the energy and momentum of the two muons.

### Key Variables

- **E1, E2** – Energy of muon 1 and muon 2  
- **px1, py1, pz1** – Momentum components of muon 1  
- **px2, py2, pz2** – Momentum components of muon 2  
- **M** – Invariant mass of the muon pair  

The invariant mass is a crucial quantity in particle physics because it allows us to identify the parent particle that decayed into the muons.

## Invariant Mass Distribution

The invariant mass of a muon pair is one of the most important quantities in this analysis. It helps identify the presence of intermediate particles that decay into two muons.

If a particle such as the Z boson is present, we expect to see a peak around its mass (~91 GeV) in the distribution.

```python
import pandas as pd
import matplotlib.pyplot as plt

# load data
df = pd.read_csv("Zmumu.csv")

# plot
plt.hist(df["M"], bins=50)
plt.title("Dimuon Invariant Mass Distribution")
plt.xlabel("Mass (GeV)")
plt.ylabel("Number of Events")
plt.show()
```

### Invariant Mass Distribution of Dimuon Events

The histogram represents the distribution of invariant mass values of dimuon events.

A dimuon event refers to a particle collision in which two muons are detected. Muons are elementary particles similar to electrons but with greater mass, and they are commonly produced in high-energy collisions at CERN’s Large Hadron Collider. Each recorded event corresponds to a single collision where two muons are identified.

For each event, properties such as the energy and momentum of the muons are measured. Using these values, the invariant mass (M) of the muon pair is calculated. This quantity represents the effective combined mass of the two muons and is a key parameter in particle physics analysis because it helps identify the parent particle that produced them.

### Interpretation


From the histogram, we observe a strong peak around 90 GeV, indicating that many dimuon events have a combined mass close to this value. This peak closely matches the known mass of the Z boson (~91 GeV), suggesting that a large number of these events are produced from Z boson decays into muon pairs.

Most of the data is concentrated between approximately 80 GeV and 100 GeV, with fewer events at lower (below 50 GeV) and higher mass values (above 110 GeV). This concentration around a specific mass range indicates the presence of a dominant physical process, namely the production and decay of the Z boson.

Overall, the histogram provides clear evidence that the dataset is not randomly distributed, but instead reflects underlying physical laws governing particle interactions and decay processes.

## Energy Relationship Between Muons

```python
plt.scatter(df["E1"], df["E2"], alpha=0.5)
plt.title("Muon Energy Comparison")
plt.xlabel("E1")
plt.ylabel("E2")
plt.show()
```

## Energy Relationship Between Muons

### Explanation

This scatter plot compares the energies of the two muons in each dimuon event. Each point represents a single event, where the x-axis shows the energy of the first muon (E1) and the y-axis shows the energy of the second muon (E2).

### Observation

The data points are widely distributed, with most values concentrated in the lower energy range (approximately 0 to 150 GeV). A smaller number of events appear at higher energy values, indicating the presence of rare high-energy events .

While the points do not form a strict linear pattern, there is a broad spread that suggests variability in how energy is shared between the two muons.

### Interpretation

The absence of a tight linear relationship indicates that the energies of the two muons are not fixed or identical, but instead vary from event to event. However, this does not mean they are completely independent.

In many cases, the muons are produced from the decay of a single parent particle, such as the Z boson, and therefore share the available energy. The variation in the scatter plot reflects differences in collision conditions and the distribution of energy between the two muons.

Overall, the plot suggests that while the energies are not strongly correlated, they are still influenced by the same underlying physical process, leading to a loosely related but variable energy distribution.

# Momentum Relationship Between Muons 

```python
plt.hist(df["pt1"], bins=50, alpha=0.5, label="Muon 1")
plt.hist(df["pt2"], bins=50, alpha=0.5, label="Muon 2")

plt.legend()
plt.title("Transverse Momentum Distribution")
plt.xlabel("pt (GeV)")
plt.ylabel("Number of Events")

plt.show()
```

## Transverse Momentum Distribution

### Explanation

This histogram compares the transverse momentum (pt) of the two muons in each dimuon event. Transverse momentum represents the component of a particle’s motion perpendicular to the beam direction and is an important quantity in collider physics, as it is less affected by the initial motion of the colliding particles.

### Observation

The majority of the transverse momentum values for both muons are concentrated between approximately 20 GeV and 60 GeV. The distributions for Muon 1 and Muon 2 are very similar and largely overlap, indicating a consistent pattern across events. Only a small number of events exhibit higher momentum values above 80 GeV.

### Interpretation

This distribution indicates that most muons are produced with moderate transverse momentum, which is typical for particles resulting from high-energy collision processes. The small number of high-momentum events suggests that such extreme conditions are less common.

The strong similarity between the two distributions reflects the fact that the muons are produced together in the same event, often from the decay of a single parent particle such as the Z boson. As a result, their momentum characteristics are closely related, leading to similar transverse momentum distributions.

Overall, the plot suggests that dimuon production follows consistent physical patterns governed by conservation laws and the underlying particle decay processes.
